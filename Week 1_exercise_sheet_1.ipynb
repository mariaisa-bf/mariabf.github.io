The cell below is an automated test for Exercise 1.  Note that passing an automated test does not guarantee that your code is correct.

```python
def run_test(test, expected, message=None):
    def coloured_print(text, colour):
        from IPython.core.display import display, HTML
        display(HTML(f'<span style=color:{colour}><pre>{text}</pre></span>'))

    exception = None
    try:
        from io import StringIO
        import sys

        old_stdout = sys.stdout
        sys.stdout = StringIO()
        exec('global must_be_true\n' + test)
        output = sys.stdout.getvalue().rstrip()
    except Exception as ex:
        exception = ex
    finally:
        sys.stdout = old_stdout

    print('-------------------------------------')
    print('Test code:')
    for line in test.split('\n'):
        print('  ' + line)

    if exception:
        coloured_print('FAIL', 'red')
        print(f'  Exception: {exception}')
        if message is not None:
            print(f'  {message}')
    else:
        passed = output == expected
        colour = 'green' if passed else 'red'
        coloured_print('PASS' if passed else 'FAIL', colour)
        coloured_print('  Expected: "' + expected + '"', colour)
        coloured_print('  Got     : "' + output + '"', colour)
        if not passed and message is not None:
            coloured_print(f'  Hint: {message}', colour)

    print()

def single_test(test, expected, message=None):
    try:
        run_test(test, expected, message)
    except:
        import traceback
        traceback.print_exc()


single_test('''print(listOfNames)''', "['Kermit the Frog', 'Fozzie Bear', 'Beaker', 'Animal', 'Miss Piggy']")
```
